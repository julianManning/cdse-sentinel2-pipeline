## 🛠️ Conda Environment Lifecycle Manager
> ### **Environment Synchronisation → Production‑Ready Reproducibility**

This module implements an **automated infrastructure‑as‑code workflow** for managing Python environments. By coupling Conda’s CLI with a custom **threading‑based monitoring engine**, the system ensures the local execution environment matches project dependency definitions without manual terminal intervention. It provides a non‑destructive **state orchestration** path that performs **delta‑based installations** and **asynchronous logging** to capture Conda output without blocking the main execution thread.

The tool is engineered for **reproducibility and stability**, providing a non‑destructive synchronisation path that updates existing environments while maintaining a live, phase‑aware progress feedback loop.

---

### 📂 Logic & Infrastructure Components
* **State Orchestration**: Validates environment existence and triggers delta‑based installations to match YAML definitions.
* **Asynchronous Logging**: Utilises a threading-based engine to capture Conda output in real-time without blocking the main execution thread.
* **Non‑Destructive Delta**: Performs selective installations where only missing packages are added, ensuring existing environments are never removed or destroyed.

---

### 📋 Behavior & Limitations
* **Selective Installation**: Installs only missing packages; existing environments are maintained to ensure system stability.
* **Conda-Only Scope**: Synchronisation logic focuses exclusively on Conda dependencies; `pip:` sections within YAML files are currently ignored.
* **Note:** It is highly recommended to restart your machine after installation ⚠️. This ensures the Windows environment variables for GDAL are properly registered.
---

### 📋 Supported Environment Definitions
* 🔒 **`env.lock.yml`**: Fully version‑pinned manifest. Guarantees 1:1 reproducibility across different machines **best for production, and CI**.
* 🔓 **`env.base.yml`**: Loose dependency specifications. Allows Conda to resolve the latest compatible versions **best for development and exploration**.

## 🔍 Acolite Configuration Integrity Check
> ### **Environment & Kernel Validation → Final Deployment Assurance**

This cell implements a **multi‑point diagnostic audit** to verify the structural integrity of the Acolite processing environment. By programmatically inspecting the Jupyter `KernelSpecManager` and cross‑referencing it with the local Conda registry, the script ensures that the "acolite" kernel is correctly mapped to a functional Python executable.

The tool acts as a **final quality gate**, preventing execution errors caused by "ghost kernels".

---

### 📂 Logic & Diagnostic Components
| Component | Responsibility |
|:---|:---|
| **Kernel Registry Audit** | **Specification Discovery** — Confirms "acolite" is a registered option in the notebook interface. |
| **JSON Schema Validation** | **Path Verification** — Parses `kernel.json` to extract the absolute path of the Python executable. |
| **Conda State Sync** | **Environment Integrity** — Validates that the underlying Conda environment actually exists on disk. |
| **Executable Mapping** | **Binary Confirmation** — Identifies the specific Python interpreter bound to the kernel. |

---

### 📋 Diagnostic Troubleshooting
If this check fails, it typically points to:
* **Kernel Not Found**: Environment exists, but registration was skipped.
* **Env Not Found**: Kernel specification exists, but the Conda environment was removed or renamed.
* **JSON Error**: The kernel path is corrupted or permissions are restricted.

```python
run_integrity_check(env_name="sentinel2")
```
